```python
# Manejo de errores

lista_numeros = []

while True:
    try:
        # 1. El programa pide un número y la conversión está dentro del 'try'.
        numero_str = input("Ingrese un número (ingrese 0 para salir): ")
        numero = int(numero_str)

        # 2. Si el número es 0, salimos del bucle con 'break'.
        if numero == 0:
            break
        
        # 3. Si no es 0, lo agregamos a la lista.
        lista_numeros.append(numero)
    
    except ValueError:
        # 4. Si hay un error, el programa muestra un mensaje y el bucle continúa.
        print(f"Error: '{numero_str}' no es un número entero válido. Inténtelo de nuevo.")

# 5. Cuando el bucle termina, el programa imprime la lista.
print(f"Los números ingresados fueron: {lista_numeros}")
```

12. Cuenta cuántas veces aparece la letra “a” en una frase ingresada, usando un `for`. 

```python
frase = input("Ingrese una frase: ")
mi_letra = "a"
contador = 0 

for letras in frase:
   if letras == mi_letra:
      contador = contador + 1

print(f"La letra '{mi_letra}' en la frase '{frase}' aparece: {contador} veces")
```

14. Contador de múltiplos de 3:
Pide números al usuario hasta que ingrese 0. Cuenta cuántos de los números ingresados son múltiplos de 3.

```python
# iniciamos con el contador
contador = 0
lista = []
lista_multiplos = []

#pedimos al usuario que ingrese un primer numero
#numero = float(input("ingrese un numero(para salir ingrese cero 0): "))

#creamos la condicional de bucle while
while True:
    
    numero = float(input("ingrese un numero(para salir ingrese cero 0): "))
    if numero == 0:
        break
    lista.append(numero)

    if numero % 3 == 0:
        contador += 1
        lista_multiplos.append(numero)

print(f"Los numeros ingresados son: {lista},\nlos numeros multiplos de 3 son:{lista_multiplos}\ny el total de multiplos de 3 son: {contador}")
```

16. Agregar elementos a una lista:
Pide palabras al usuario y agrégalas a una lista hasta que escriba "fin". Muestra la lista final.

```python
#iniciamos con la lista donde vamos a guardar las palabras
lista = []

#peticion = input("ingrese palabras(Para salir escriba 'fin'): ")

#comenzamos las condiciones con bucle while
while True:

    peticion = input("ingrese palabras(Para salir escriba 'fin'): ").lower() # lower() para que acepte cualquier "fin"
    if peticion == "fin":   #si ingresa "fin" el bucle se detiene
        break

    else:

        lista.append(peticion) #si no, guarda las palabras en una lista

print(f"Las palabras ingresadas, fueron: {lista}") #imprime el resultado
    
```
