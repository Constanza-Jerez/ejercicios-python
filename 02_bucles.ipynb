```python
# Manejo de errores

lista_numeros = []

while True:
    try:
        # 1. El programa pide un número y la conversión está dentro del 'try'.
        numero_str = input("Ingrese un número (ingrese 0 para salir): ")
        numero = int(numero_str)

        # 2. Si el número es 0, salimos del bucle con 'break'.
        if numero == 0:
            break
        
        # 3. Si no es 0, lo agregamos a la lista.
        lista_numeros.append(numero)
    
    except ValueError:
        # 4. Si hay un error, el programa muestra un mensaje y el bucle continúa.
        print(f"Error: '{numero_str}' no es un número entero válido. Inténtelo de nuevo.")

# 5. Cuando el bucle termina, el programa imprime la lista.
print(f"Los números ingresados fueron: {lista_numeros}")
```

15. Promedio de N números:
Pide al usuario cuántos números va a ingresar. Usa un while para pedir cada número y calcula el promedio al final.

```python
#promedio = suma / cantidad
#suma = acumulador --> acumulador
#contador --> cantidad

cantidad = 0     # cuantos numeros hemos pedido
acumulador = 0   # donde sumaremos todos los numeros ingresados
lista_numeros = []

#condiciones en bucle while para pedir cada numero uno por uno y las sumamos, no tiene limite
while True:

    numero = int(input("Ingrese numeros: "))

    if numero == 0:  # condicion if de corte del bucle
        break

    lista_numeros.append(numero) #guardamos los numeros ingresados en una lista

    cantidad = cantidad + 1 #cada vez que cuento es 1 en 1
    acumulador = acumulador + numero #suma de los numeros ingresados

    promedio = acumulador / cantidad

print(f"El promedio de los números {lista_numeros} es: {promedio}")
```

19. Contar vocales:
Pide una palabra al usuario y usa un while para contar cuántas vocales tiene.

```python
cadena = "Hola Mundo"
contador = 0
indice = 0
vocales = "aeiouAEIOU"

while indice < len(cadena):
    if cadena[indice] in vocales:
        contador = contador +1
    indice = indice +1

print(f"El texto '{cadena}' tiene {contador} vocales")
```

```python
texto = input("Ingrese una palabra: ").lower() #inicio solicitando una palabra
contador_vocales = 0                           # contador de vocales 1 a 1  
indice = 0                                     # iniciar un indice para recorrer la cadena, comenzando por el 1er caracter
vocales = "aeiou"                              # se definen las vocales

while indice < len(texto):                     #mientras indice sea menor a la longitud del texto
    if texto[indice] in vocales:               #si en cada iteracion hay una vocal se suma en el contador
        contador_vocales += 1
    indice += 1                                #sumamos 1 al indice para que pase a la sgte iteracion
    
print(f"La palabra '{texto}' tiene {contador_vocales} vocales.")
```

20. Invertir un string:
Pide una palabra al usuario y usa un while para mostrarla al revés.

```python
texto = "Hola Mundo!"
texto_invertido = texto[::-1] #iterable[inicio:fin:paso]
print(f"El texto invertido es {texto_invertido}")

#Cuando [::-1] se utiliza, la sintaxis se interpreta de la siguiente manera: 
# : (primer colon):
# Se omite el valor de inicio. Esto indica que el slicing comenzará desde el principio del iterable.
# : (segundo colon):
# Se omite el valor de fin. Esto indica que el slicing llegará hasta el final del iterable.
# -1 (paso):
# Se especifica un paso de -1. Esto significa que los elementos se recorrerán hacia atrás, es decir, del último al primero.
```

```python
texto = input("Ingrese una palabra")
texto_invertido = []
indice = len(texto) #largo del texto

while indice > 0: #mientras el largo de texto sea mayor a 0
    texto_invertido.append(texto[indice - 1]) #la indexacion del string // guardar los elementos de la palabra o frase
    indice -=1 #para acceder a todos los elementos de la lista

print(f"La lista reversada es '{''.join(texto_invertido)}'") #con join uno los caracteres del texto
```
